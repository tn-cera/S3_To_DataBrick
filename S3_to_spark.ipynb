```python
sales = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("s3://")

display(sales)
```

```python
customers = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("s3://")
)

customers.show()
```

```python
result=(sales.filter(sales.amount>500).join(customers,"customer_id").groupby("city").sum("amount"))
```

```python
display(result)
```

```python
result.write.csv("s3://")
```
